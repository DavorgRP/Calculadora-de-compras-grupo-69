```python
from IPython.display import clear_output, display
import ipywidgets as widgets


titulo = widgets.HTML("<h2 style='color: #2B548A;'>🧮 CALCULADORA DE COMPRA</h2>")


txt_producto = widgets.Text(
    description="Producto:",
    placeholder="Ej: Cuadernos, Manzanas..."
)

num_precio = widgets.FloatText(
    description="Precio (Bs):",
    value=0.0,
    step=0.50
)

num_cantidad = widgets.IntText(
    description="Cantidad:",
    value=1
)


btn_calcular = widgets.Button(
    description="Calcular Total",
    button_style="success",
    icon="calculator"
)

btn_limpiar = widgets.Button(
    description="Limpiar",
    button_style="warning",
    icon="refresh"
)


salida = widgets.Output()



def calcular_compra(b):
    with salida:
        clear_output(wait=True)

        producto = txt_producto.value.strip()
        precio = num_precio.value
        cantidad = num_cantidad.value


        if not producto:
            print("⚠️ Error: Debe escribir el nombre del producto.")
            return

        if precio <= 0:
            print("⚠️ Error: El precio unitario debe ser mayor a cero.")
            return

        if cantidad <= 0:
            print("⚠️ Error: La cantidad debe ser de al menos 1 unidad.")
            return


        total = precio * cantidad


        resumen_html = f"""
        <div style="background-color: #f4f6f9; padding: 15px; border-radius: 8px; border-left: 5px solid #2B548A; max-width: 420px; font-family: sans-serif; margin-top: 10px;">
            <h3 style="margin-top: 0; color: #2B548A;">🛍️ RESUMEN DE LA COMPRA</h3>
            <p style="margin: 5px 0;"><b>• Producto:</b> {producto.title()}</p>
            <p style="margin: 5px 0;"><b>• Precio Unitario:</b> BS {precio:,.2f}</p>
            <p style="margin: 5px 0;"><b>• Cantidad:</b> {cantidad} unidades</p>
            <hr style="border: 0; border-top: 1px solid #ccc; margin: 10px 0;">
            <h3 style="color: #2e7d32; margin: 0;">💰 TOTAL A PAGAR: Bs {total:,.2f}</h3>
        </div>
        """
        display(widgets.HTML(resumen_html))



def limpiar_campos(b):
    txt_producto.value = ""
    num_precio.value = 0.0
    num_cantidad.value = 1
    with salida:
        clear_output(wait=True)



btn_calcular.on_click(calcular_compra)
btn_limpiar.on_click(limpiar_campos)


panel_botones = widgets.HBox([btn_calcular, btn_limpiar])
display(titulo, txt_producto, num_precio, num_cantidad, panel_botones, salida)
```
